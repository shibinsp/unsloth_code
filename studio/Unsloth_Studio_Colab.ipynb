```python
# Unsloth Studio — robust Colab launcher (replaces the iframe cell)
import sys
import socket
import requests
from IPython.display import HTML, display
from google.colab import output

PORT = 8888
HOST = "127.0.0.1"
TIMEOUT = 5

def _port_open(host: str, port: int, timeout: float = 2.0) -> bool:
    with socket.socket(socket.AF_INET, socket.SOCK_STREAM) as sock:
        sock.settimeout(timeout)
        return sock.connect_ex((host, port)) == 0

def _http_status(url: str, timeout: float = TIMEOUT) -> int:
    try:
        return requests.get(url, timeout=timeout, allow_redirects=False).status_code
    except requests.RequestException:
        return 0

# 1. Verify the backend is reachable on loopback
if not _port_open(HOST, PORT):
    sys.exit(f"❌ Nothing listening on {HOST}:{PORT}. Re-run the Start Unsloth Studio cell.")

status = _http_status(f"http://{HOST}:{PORT}")
if status == 0:
    sys.exit(f"❌ TCP open but HTTP unreachable on {HOST}:{PORT}.")
print(f"✅ Backend healthy on {HOST}:{PORT}  (HTTP {status})")

# 2. Get the proxied URL (top-level — works even when iframe is blocked by 3p-cookie policy)
proxied_url = output.eval_js(
    f'google.colab.kernel.proxyPort({PORT}, {{"cache": false}})',
    timeout_sec=30,
)
print(f"🔗 Proxied URL: {proxied_url}")

# 3. Render an "Open" button + try the iframe; if the browser blocks it, the button still works
display(HTML(f"""
<div style="font-family:system-ui,-apple-system,sans-serif;padding:12px 16px;
            border:1px solid #e0e0e0;border-radius:10px;background:#fafafa;margin:8px 0;">
  <div style="display:flex;align-items:center;gap:12px;margin-bottom:8px;">
    <span style="font-size:16px;font-weight:600;">🦥 Unsloth Studio is Ready</span>
    <a href="{proxied_url}" target="_blank" rel="noopener"
       style="background:#111;color:#fff;text-decoration:none;padding:8px 14px;
              border-radius:6px;font-size:14px;font-weight:500;">▶ Open in new tab</a>
  </div>
  <div style="font-size:12px;color:#666;">
    Inline preview below (may be blocked by third-party cookie settings — use the button if blank).
  </div>
</div>
<iframe src="{proxied_url}" width="100%" height="900"
        style="border:1px solid #e0e0e0;border-radius:8px;"
        referrerpolicy="no-referrer-when-downgrade"
        allow="clipboard-read; clipboard-write"></iframe>
"""))
```
